```python
from copy import deepcopy
from datetime import datetime
from warnings import warn

inova_metadata_df = pd.read_excel(INOVA_METADATA_PATH, header=[0,1])
inova_metadata_df.columns = [
    f"{col[0]}_{col[1]}".strip().lower().replace(" ", "_")
    if col[0] and "time" in str(col[0]).lower()
    else str(col[1]).strip().lower().replace(" ", "_")
    for col in inova_metadata_df.columns
]

# columns that are *supposed* to be numeric but might need repairs
# TODO: add all numeric covariate columns here just for safety
numeric_cols_with_some_non_numeric = [
    'timepoint_1_height_(cm)', 'timepoint_1_height_%ile', 'timepoint_1_height_zscore', 'timepoint_1_weight_zscore', 'timepoint_1_bmi_(kg/m^2)', 'timepoint_1_bmi_%ile', 'timepoint_1_bmi_zscore', 'timepoint_1_wbc_(k/mcl)', 'timepoint_1_crp_(mg/l)', 'timepoint_1_calprotectin', 'timepoint_1_alt', 'timepoint_1_tbili', 'timepoint_1_dbili', 'timepoint_1_folate', 
    'timepoint_2_wbc_(k/mcl)', 'timepoint_2_hematocrit_(%)', 'timepoint_2_albumin_(gm/dl)', 'timepoint_2_crp_(mg/l)', 'timepoint_2_calprotectin', 'timepoint_2_folate', 'timepoint_2_tbili', 'timepoint_2_drug_trough_level', 'timepoint_2_antibody_level', 
    'timepoint_3_albumin_(gm/dl)', 'timepoint_3_crp_(mg/l)', 'timepoint_3_calprotectin', 'timepoint_3_tbili', 'timepoint_3_dbili', 'timepoint_3_drug_trough_level', 'timepoint_3_antibody_level', 'timepoint_3_folate', 
    'timepoint_4_weight_(kg)', 'timepoint_4_hematocrit_(%)', 'timepoint_4_esr_(mm/hr)', 'timepoint_4_crp_(mg/l)', 'timepoint_4_tbili', 'timepoint_4_drug_trough_level', 'timepoint_4_antibody_level', 'timepoint_4_folate',
    'timepoint_5_wbc_(k/mcl)', 'timepoint_5_esr_(mm/hr)', 'timepoint_5_crp_(mg/l)', 'timepoint_5_calprotectin', 'timepoint_5_tbili', 'timepoint_5_drug_trough_level', 'timepoint_5_antibody_level', 'timepoint_5_folate',
    'timepoint_6_weight_(kg)', 'timepoint_6_wbc_(k/mcl)', 'timepoint_6_hemoglobin_(gm/dl)', 'timepoint_6_hematocrit_(%)', 'timepoint_6_crp_(mg/l)', 'timepoint_6_25-hydroxy_vitamin_d', 'timepoint_6_ferritin_(ng/ml)', 'timepoint_6_folate', 'timepoint_6_alt', 'timepoint_6_tbili', 'timepoint_6_drug_trough_level', 'timepoint_6_antibody_level', 
    'timepoint_7_height_(cm)', 'timepoint_7_bmi_(kg/m^2)', 'timepoint_7_albumin_(gm/dl)', 'timepoint_7_crp_(mg/l)', 'timepoint_7_tbili', 'timepoint_7_drug_trough_level', 'timepoint_7_antibody_level', 'timepoint_7_folate',
    'timepoint_8_bmi_(kg/m^2)', 'timepoint_8_crp_(mg/l)', 'timepoint_8_tbili', 'timepoint_8_antibody_level', 'timepoint_8_dose_(mg)', 'timepoint_8_folate',
    'timepoint_9_esr_(mm/hr)', 'timepoint_9_crp_(mg/l)', 'timepoint_9_calprotectin', 'timepoint_9_antibody_level', 'timepoint_9_folate',
    'timepoint_10_tbili', 'timepoint_10_antibody_level', 'timepoint_10_crp_(mg/l)', 'timepoint_10_folate',
    'timepoint_11_crp_(mg/l)', 'timepoint_11_folate',
    'timepoint_12_crp_(mg/l)', 'timepoint_12_folate'
]

print(numeric_cols_with_some_non_numeric)

for col in numeric_cols_with_some_non_numeric:
    values_in_column = inova_metadata_df[col].tolist()
    updated_values = deepcopy(values_in_column)
    for i, value in enumerate(values_in_column):
        if isinstance(value, (int, float)): continue
        if not isinstance(value, str):
            warn(f"Failed to sensibly coerce {value} into a numeric datatype, falling back on 'None' (treating this value as missing)")
            updated_values[i] = None
            continue
        # for antibody levels where the ONLY text is 'not detected' (no LOD provided)
        if value == 'not detected':
            updated_values[i] = 0
            continue
        else:
            # if a value is something like <0.1, treat it as 0.1 or >0.3 treat it as 0.3
            # interpret commas as decimals (only if there are no decimal points already present): 17,6 -> 17.6
            # when an antibody level is below LOD, it may be recorded as <22, not detected (or similar). Coerce this to -> 22
            updated_value = value.strip("<>").replace(", not detected", "")
            if "," in updated_value and "." not in updated_value:
                updated_value = updated_value.replace(",", ".")
            
            try:
                float(updated_value)
            except ValueError:
                warn(f"Failed to sensibly coerce {value} into a numeric datatype (attempted with converting to {updated_value}), falling back on 'None' (treating this value as missing)")
                updated_value = None

        assert updated_value is None or isinstance(float(updated_value), float), f"{value} was improperly coerced to {updated_value}"
        updated_values[i] = updated_value

    inova_metadata_df[col] = updated_values
    inova_metadata_df[col] = inova_metadata_df[col].astype(float)

inova_metadata_df = inova_metadata_df.replace("Unchecked", False)
inova_metadata_df = inova_metadata_df.replace("Checked", True)     
inova_metadata_df
```

```python
from math import log

BMI_REFERENCE_PATH = Path.cwd() / "data" / "bmiagerev.csv"
WAZ_REFERENCE_PATH = Path.cwd() / "data" / "wtage.csv"
STATURE_REFERENCE_PATH = Path.cwd() / "data" / "statage.csv"

bmi_reference_df = pd.read_csv(BMI_REFERENCE_PATH, skiprows=[220])
waz_reference_df = pd.read_csv(WAZ_REFERENCE_PATH)
stature_reference_df = pd.read_csv(STATURE_REFERENCE_PATH)

def get_interpolated_lms(sex : int, age_in_months : float, reference_df : pd.DataFrame) -> (float, float, float):
    assert sex == 1 or sex == 2
    assert 24 <= age_in_months <= 240.5
    reference_for_sex = reference_df[reference_df['Sex'] == sex]
    nearest_younger_than_requested = reference_for_sex[reference_for_sex['Agemos'].le(age_in_months)].sort_values(by='Agemos', ascending=False).head(1)
    nearest_older_than_requested = reference_for_sex[reference_for_sex['Agemos'].ge(age_in_months)].sort_values(by='Agemos', ascending=True).head(1)

    reference_to_interpolate = pd.concat([nearest_younger_than_requested, nearest_older_than_requested]).sort_values(by='Agemos', ascending=True)
    assert len(reference_to_interpolate) == 2, f"Interpolating between {reference_to_interpolate}"
    scale_lower = reference_to_interpolate['Agemos'].iloc[0]
    scale_upper = reference_to_interpolate['Agemos'].iloc[-1]
    assert scale_lower <= age_in_months <= scale_upper
    # dirty hack if scale upper and scale lower are the same
    linear_factor = (age_in_months - scale_lower) / (scale_upper - scale_lower + 1e-9)

    L_interpolated = reference_to_interpolate['L'].iloc[0] * linear_factor + reference_to_interpolate['L'].iloc[1] * (1 - linear_factor)
    M_interpolated = reference_to_interpolate['M'].iloc[0] * linear_factor + reference_to_interpolate['M'].iloc[1] * (1 - linear_factor)
    S_interpolated = reference_to_interpolate['S'].iloc[0] * linear_factor + reference_to_interpolate['S'].iloc[1] * (1 - linear_factor)

    return L_interpolated, M_interpolated, S_interpolated

# anthropometrics
def calculate_waz(sex : int, age_in_months : float, weight_kg : float) -> float:
    L,M,S = get_interpolated_lms(sex, age_in_months, waz_reference_df)
    return ((weight_kg / M)**L - 1) / (L * S) if L != 0 else log(weight_kg / M) / S
    
def calculate_haz(sex : int, age_in_months : float, height_cm : float) -> float:
    L,M,S = get_interpolated_lms(sex, age_in_months, stature_reference_df)
    return ((height_cm / M)**L - 1) / (L * S) if L != 0 else log(height_cm / M) / S
    
def calculate_baz(sex : int, age_in_months : float, bmi : float) -> float:
    L,M,S = get_interpolated_lms(sex, age_in_months, bmi_reference_df)
    return ((bmi / M)**L - 1) / (L * S) if L != 0 else log(bmi / M) / S
```

```python
def process_survival_row(row : dict):
    id = row['study_id']
    assert row['init_biologic_specified'] == 1 or row['init_biologic_specified'] == 2, f"Biologic specified was {row['init_biologic_specified']} for {id}"
    age_at_anti_tnf_induction = np.nanmin([row['age_at_induction_date'], row['age_calculated'] + row['diagnosis_induction_date_calc'] / 12.])
    # drug switch between dx and fu1 to a non-anti-tnf?
    if row['change_intherapyyn'] == 1 and row['change_intherapydrug'] not in [1, 2]:
        age_at_switch = row['age_at_change_intherapy_date']
        # TODO: ask Ethan/Samir, are we not recording reason for switch if switched between dx and fu1?
        event = 1
        return id, 0, age_at_switch - age_at_anti_tnf_induction, event
            
    # drug switch between fu1 and fu2?
    if row['is_therapychange'] == 1 and row['new_drug_one'] not in [1, 2]:
        age_at_switch = np.nanmin([row['age_at_therapychange_date_one'], row['age_at_induction_start_date_one']])
        # switched due to insurance/preference
        event = not (row['therapychange_ivl_reason'] == 1 or row['therapychange_ivl_reason'] == 3)
        return id, 0, age_at_switch - age_at_anti_tnf_induction, event

    # never switched
    all_dates = ['age_at_dob', 'age_at_date_enrolled', 'age_at_radio_date', 'age_at_date_officevis', 'age_at_weight_date', 'age_at_height_date', 'age_at_date_of_blood_draw_v2', 'age_at_cmp_date', 'age_at_calprotectin_date', 'age_at_induction_date', 'age_at_dose_date', 'age_at_fu_date', 'age_at_change_intherapy_date', 'age_at_fu_height_date', 'age_at_fu_date_of_blood_draw', 'age_at_fu_cmp_date', 'age_at_fu_calprotectin_date', 'age_at_fu_colonoscopy_date', 'age_at_yr_fu_date', 'age_at_interval_scope_date', 'age_at_therapychange_date_one', 'age_at_previous_anser_date_one', 'age_at_induction_start_date_one', 'age_at_maintenance_start_date_one', 'age_at_new_anser_date_one', 'age_at_dosechange_date_one', 'age_at_fiveyr_weight_date', 'age_at_fiveyr_date_of_blood_draw', 'age_at_fiveyr_cmp_date', 'age_at_fiveyr_calprotectin_date']
    age_at_last_known_visit = np.nanmax([
        row[date] for date in all_dates
    ])
    event = 0
    return id, 0, age_at_last_known_visit - age_at_anti_tnf_induction, event
```

```python
import warnings
import lifelines

didnt_start_on_anti_tnf = ~duke_metadata_df['init_biologic_specified'].isin([1,2])
if didnt_start_on_anti_tnf.any():
    warnings.warn(f"The following patients didn't start on Anti-TNF! {duke_metadata_df[didnt_start_on_anti_tnf]['study_id']}")
    
duke_metadata_df = duke_metadata_df[duke_metadata_df['init_biologic_specified'].isin([1,2])]
res = duke_metadata_df.apply(process_survival_row, axis=1)
duke_survival_df = pd.DataFrame(res.to_list(), columns=['id', 'start', 'stop', 'event'])
duke_survival_df['event'].value_counts()
```

```python
from lifelines.utils import add_covariate_to_timeline

def add_covariates_to_duke_survival_df(base_survival_df : pd.DataFrame, full_df : pd.DataFrame) -> pd.DataFrame:
    df_with_covariates = base_survival_df.copy()

    # Complete Blood Count (CBC)
    
    cbc_cvs = {
        'hgb',
        'hct',
        'plateletcount',
        'wbc',
        'neutrophils',
        'lymphocytes',
        'eosinophils',
    }

    for cv in cbc_cvs:
        cv_dict = {'id': [], cv: [], 'duration': []}
        for row in full_df.to_dict(orient='records'):
            id = row['study_id']
            # dx
            if row['cbc_yn'] == 1:
                time_to_measurement = row['age_at_date_of_blood_draw_v2'] - row['age_at_induction_date']
                measured_value = row[cv]
                cv_dict['id'].append(id)
                cv_dict[cv].append(measured_value)
                cv_dict['duration'].append(time_to_measurement)
            
            # fu1
            if row['fu_cbc_yn'] == 1:
                time_to_measurement = row['age_at_fu_date_of_blood_draw'] - row['age_at_induction_date']
                measured_value = row[f'fu_{cv}']
                cv_dict['id'].append(id)
                cv_dict[cv].append(measured_value)
                cv_dict['duration'].append(time_to_measurement)
                
            # fu2
            if row['fiveyr_cbc_yn'] == 1:
                time_to_measurement = row['age_at_fiveyr_date_of_blood_draw'] - row['age_at_induction_date']
                measured_value = row[f'fiveyr_{cv}']
                cv_dict['id'].append(id)
                cv_dict[cv].append(measured_value)
                cv_dict['duration'].append(time_to_measurement)
        cv_df = pd.DataFrame(cv_dict)
        df_with_covariates = add_covariate_to_timeline(df_with_covariates, cv_df, id_col = 'id', duration_col = 'duration', event_col = 'event')

    # Comprehensive Metabolic Panel (CMP)
    
    cmp_cvs = {
        'albumin',
        'urea_calculated',
        'creatinine',
        'ast',
        'alt',
        'alp',
    }

    for cv in cmp_cvs:
        cv_dict = {'id': [], cv: [], 'duration': []}
        for row in full_df.to_dict(orient='records'):
            id = row['study_id']
            # dx
            if row['cmp_yn'] == 1:
                time_to_measurement = row['age_at_cmp_date'] - row['age_at_induction_date']
                measured_value = row[cv]
                cv_dict['id'].append(id)
                cv_dict[cv].append(measured_value)
                cv_dict['duration'].append(time_to_measurement)
            
            # fu1
            if row['fu_cmp_yn'] == 1:
                time_to_measurement = row['age_at_fu_cmp_date'] - row['age_at_induction_date']
                measured_value = row[f'fu_{cv}']
                cv_dict['id'].append(id)
                cv_dict[cv].append(measured_value)
                cv_dict['duration'].append(time_to_measurement)
                
            # fu2
            if row['fiveyr_cmp_yn'] == 1:
                time_to_measurement = row['age_at_fiveyr_cmp_date'] - row['age_at_induction_date']
                measured_value = row[f'fiveyr_{cv}']
                cv_dict['id'].append(id)
                cv_dict[cv].append(measured_value)
                cv_dict['duration'].append(time_to_measurement)
        cv_df = pd.DataFrame(cv_dict)
        df_with_covariates = add_covariate_to_timeline(df_with_covariates, cv_df, id_col = 'id', duration_col = 'duration', event_col = 'event')

    # ESR
    cv_dict = {'id': [], 'esr': [], 'duration': []}
    for row in full_df.to_dict(orient='records'):
        id = row['study_id']
        # dx
        if row['esr_yn'] == 1:
            time_to_measurement = row['age_at_date_of_blood_draw_v2'] - row['age_at_induction_date']
            measured_value = row['esr']
            cv_dict['id'].append(id)
            cv_dict['esr'].append(measured_value)
            cv_dict['duration'].append(time_to_measurement)
        
        # fu1
        if row['fu_esr_yn'] == 1:
            time_to_measurement = row['age_at_fu_date_of_blood_draw'] - row['age_at_induction_date']
            measured_value = row['fu_esr']
            cv_dict['id'].append(id)
            cv_dict['esr'].append(measured_value)
            cv_dict['duration'].append(time_to_measurement)

        if row['fiveyr_esr_yn'] == 1:
            time_to_measurement = row['age_at_fiveyr_date_of_blood_draw'] - row['age_at_induction_date']
            measured_value = row['fiveyr_esr']
            cv_dict['id'].append(id)
            cv_dict['esr'].append(measured_value)
            cv_dict['duration'].append(time_to_measurement)

    cv_df = pd.DataFrame(cv_dict)
    df_with_covariates = add_covariate_to_timeline(df_with_covariates, cv_df, id_col = 'id', duration_col = 'duration', event_col = 'event')
    
    # CRP
    cv_dict = {'id': [], 'crp': [], 'duration': []}
    for row in full_df.to_dict(orient='records'):
        id = row['study_id']
        # dx
        if row['crp_yn'] == 1:
            time_to_measurement = row['age_at_date_of_blood_draw_v2'] - row['age_at_induction_date']
            measured_value = row['crp']
            cv_dict['id'].append(id)
            cv_dict['crp'].append(measured_value)
            cv_dict['duration'].append(time_to_measurement)
        
        # fu1
        if row['fu_crp_yn'] == 1:
            time_to_measurement = row['age_at_fu_date_of_blood_draw'] - row['age_at_induction_date']
            measured_value = row['fu_crp']
            cv_dict['id'].append(id)
            cv_dict['crp'].append(measured_value)
            cv_dict['duration'].append(time_to_measurement)

        # CRP not measured at fiveyr fu
    cv_df = pd.DataFrame(cv_dict)
    df_with_covariates = add_covariate_to_timeline(df_with_covariates, cv_df, id_col = 'id', duration_col = 'duration', event_col = 'event')
    
    # Calprotectin
    cv_dict = {'id': [], 'calprotectin': [], 'duration': []}
    for row in full_df.to_dict(orient='records'):
        id = row['study_id']
        # dx
        if row['calpro'] == 1:
            measured_value = row['calprotectin']
            time_to_measurement = row['age_at_calprotectin_date'] - row['age_at_induction_date']
            cv_dict['id'].append(id)
            cv_dict['calprotectin'].append(measured_value)
            cv_dict['duration'].append(time_to_measurement)
            
        # fu1
        if row['fu_calpro'] == 1:
            measured_value = row['fu_calprotectin']
            time_to_measurement = row['age_at_fu_calprotectin_date'] - row['age_at_induction_date']
            cv_dict['id'].append(id)
            cv_dict['calprotectin'].append(measured_value)
            cv_dict['duration'].append(time_to_measurement)
            
        # fu2
        if row['fiveyr_calpro'] == 1:
            measured_value = row['fiveyr_calprotectin']
            time_to_measurement = row['age_at_fiveyr_calprotectin_date'] - row['age_at_induction_date']
            cv_dict['id'].append(id)
            cv_dict['calprotectin'].append(measured_value)
            cv_dict['duration'].append(time_to_measurement)

    cv_df = pd.DataFrame(cv_dict)
    df_with_covariates = add_covariate_to_timeline(df_with_covariates, cv_df, id_col = 'id', duration_col = 'duration', event_col = 'event')

    # Anthropometrics (WAZ, HAZ, WHZ, BMI)
    # TODO
    cv_dict = {'id': [], 'bmi': [], 'duration': []}
    for row in full_df.to_dict(orient='records'):
        id = row['study_id']
        # dx
        time_to_measurement = np.nanmean([row['age_at_weight_date'], row['age_at_height_date']]) - row['age_at_induction_date']
        measured_value = row['bmi']
        cv_dict['id'].append(id)
        cv_dict['bmi'].append(measured_value)
        cv_dict['duration'].append(time_to_measurement)
        
        # fu1
        # no weight date for fu1
        time_to_measurement = row['age_at_fu_height_date'] - row['age_at_induction_date']
        measured_value = row['fu_bmi']
        cv_dict['id'].append(id)
        cv_dict['bmi'].append(measured_value)
        cv_dict['duration'].append(time_to_measurement)

        # BMI can't be calculated for fu2, no height
    cv_df = pd.DataFrame(cv_dict)
    df_with_covariates = add_covariate_to_timeline(df_with_covariates, cv_df, id_col = 'id', duration_col = 'duration', event_col = 'event')

    return df_with_covariates
```

```python
from sklearn.impute import KNNImputer
from sklearn.preprocessing import RobustScaler

def clean_and_impute(df: pd.DataFrame, missing_threshold: float) -> pd.DataFrame:
    """
    1. Drops features missing more than `missing_threshold` (default 50%) of their values.
    2. Sorts by id and time to ensure correct chronological order.
    3. Forward-fills missing covariate values within each subject's id.
    """
    df = df.copy()
    structural_cols = {'id', 'start', 'stop', 'event'}
    covariate_cols = [col for col in df.columns if col not in structural_cols]
    df = df.sort_values(by=['id', 'start'])
    df[covariate_cols] = (
        df.groupby('id', group_keys=False)[covariate_cols].ffill()
    )

    min_non_nulls = len(df) * (1 - missing_threshold)
    df = df.dropna(axis=1, thresh=min_non_nulls)

    # for anything remaining, impute using KNN
    numeric_cols = df.select_dtypes(include=[np.number]).columns.to_list()
    numeric_cols = [col for col in numeric_cols if col not in structural_cols]
    X = df[numeric_cols].to_numpy()

    scaler = RobustScaler()
    imputer = KNNImputer()

    X_scaled = scaler.fit_transform(X)
    X_imputed_scaled = imputer.fit_transform(X_scaled)
    X_original_scale = scaler.inverse_transform(X_imputed_scaled)
    
    df[numeric_cols] = X_original_scale
    return df
```

```python
from lifelines.fitters.cox_time_varying_fitter import CoxTimeVaryingFitter
from sklearn.model_selection import KFold

def cross_validate(splits : int, l2_coefficient : int, l1_ratio : int) -> float:
    scores = np.zeros(splits)
    kf = KFold(n_splits=splits)
    for fold, (duke_train_idx, duke_val_idx) in enumerate(kf.split(duke_df_with_covariates['id'].unique())):
        duke_train_ids = duke_df_with_covariates['id'].unique()[duke_train_idx]
        duke_val_ids = duke_df_with_covariates['id'].unique()[duke_val_idx]
        
        duke_train = duke_df_with_covariates[duke_df_with_covariates['id'].isin(duke_train_ids)]
        duke_val = duke_df_with_covariates[duke_df_with_covariates['id'].isin(duke_val_ids)]
    
        cph = CoxTimeVaryingFitter(penalizer=l2_coefficient, l1_ratio=l1_ratio)
        cph.fit(duke_train.drop(columns='id'), 'event')
        scores[fold] = None
    return np.mean(scores)
    

coxfitter = CoxTimeVaryingFitter(penalizer=0.0)
coxfitter.fit(duke_df_with_covariates.drop(columns='id'), 'event')
coxfitter.print_summary()
coxfitter.plot()
coxfitter.log_likelihood_ratio_test()
```

```python
import numpy as np
import pandas as pd

def process_survival_row(row):
  has_ifx = not pd.isna(row['timepoints_start_ifx_dol_m'])
  has_ada = not pd.isna(row['timepoints_start_adalimumab_dol_m'])
  id = row['study_id']

  if not has_ifx and not has_ada:
    return pd.Series([id, np.nan, np.nan, np.nan, np.nan])

  starts = [s for s in [row['timepoints_start_ifx_dol_m'], row['timepoints_start_adalimumab_dol_m']] if not pd.isna(s)]
  age_at_therapy_start = min(starts)

  # Case A: IFX only
  if has_ifx and not has_ada:
    # switched off of IFX
    if not pd.isna(row['timepoints_end_ifx_dol_m']):
      return id, age_at_therapy_start, age_at_therapy_start + row[f'timepoints_duration_on_ifx_therapy_(months)__**if_switched_from_ifx->_other_biologic'], 1
    # stayed on IFX
    else:
      return (
          id,
          age_at_therapy_start,
          age_at_therapy_start + row['timepoints_duration_on_ifx_therapy_(months)__**if_did_not_change_therapy'],
          0,
      )  # Censored: stayed on IFX

  # Case B: Adalimumab only
  elif has_ada and not has_ifx:
    # switched off of ADA
    if not pd.isna(row['timepoints_end_adalimumab_dol_m']):
      return id, age_at_therapy_start, age_at_therapy_start + row[f'timepoints_duration_on_adalimumab_therapy_(months)__**if_switched_from_adalimumab_->_other_biologic'], 1
    # stayed on ADA
    else:
      return (
          id,
          age_at_therapy_start,
          age_at_therapy_start + row['timepoints_duration_on_adalimumab_therapy_(months)__**if_did_not_change_therapy'],
          0,
      )  # Censored: stayed on ADA

  # Case C: Took both
  else:
    first_drug = 'ifx' if (row['timepoints_start_ifx_dol_m'] < row['timepoints_start_adalimumab_dol_m']) else 'adalimumab'
    second_drug = 'ifx' if first_drug == 'adalimumab' else 'adalimumab'
    # switched off of anti-tnf
    if not pd.isna(row[f'timepoints_end_{second_drug}_dol_m']):
        time_on_first_drug = row[f'timepoints_duration_on_{first_drug}_therapy_(months)__**if_switched_from_{first_drug + ("_" if first_drug == "adalimumab" else "")}->_other_biologic']
        time_on_second_drug = row[f'timepoints_duration_on_{second_drug}_therapy_(months)__**if_switched_from_{second_drug + ("_" if second_drug == "adalimumab" else "")}->_other_biologic']
        return id, age_at_therapy_start, age_at_therapy_start + time_on_first_drug + time_on_second_drug, 1
    # stayed on anti-tnf
    else:
        # spreadsheet has an extra space for the ada column for some reason
        time_on_first_drug = row[f'timepoints_duration_on_{first_drug}_therapy_(months)__**if_switched_from_{first_drug + ("_" if first_drug == "adalimumab" else "")}->_other_biologic']
        time_on_second_drug = time_on_second_drug = row[f'timepoints_duration_on_{second_drug}_therapy_(months)__**if_did_not_change_therapy']
        # TODO: discuss with Eve May - what if the two intervals don't overlap?
        return id, age_at_therapy_start, age_at_therapy_start + time_on_first_drug + time_on_second_drug, 0


res = inova_metadata_df.apply(process_survival_row, axis=1)
inova_survival_df = pd.DataFrame(res.to_list(), columns=['id', 'start', 'stop', 'event'])
if np.any(inova_survival_df['stop'] < 0):
    warn(f"Found a negative duration for id(s) {inova_survival_df[inova_survival_df['stop'] < inova_survival_df['start']]['id']}. Please investigate.")

inova_survival_df = inova_survival_df[inova_survival_df['stop'] > inova_survival_df['start']]
inova_survival_df
```

```python
from lifelines.utils import add_covariate_to_timeline

def add_covariates_to_inova_survival_df(base_survival_df : pd.DataFrame, full_df : pd.DataFrame) -> pd.DataFrame:
    df_with_covariates = base_survival_df.copy()
    covariates = {
        'anthro_dol_m' : [
            'height_(cm)',
            'weight_(kg)',
            'bmi_(kg/m^2)',
        ]
        ,
        'labwork_dol_m': [
            'wbc_(k/mcl)',
            'hemoglobin_(gm/dl)',
            'hematocrit_(%)',
            'platelets_(k/mcl)',
            'albumin_(gm/dl)',
            'esr_(mm/hr)',
            'crp_(mg/l)',
            'calprotectin',
            '25-hydroxy_vitamin_d',
            'ferritin_(ng/ml)',
            'b12_(pg/ml)',
            'folate',
            'ast',
            'alt',
            'inr',
            'ggt',
            'tbili',
            'dbili',
            'drug_trough_level',
            'antibody_level'
        ]
    }

    for measured_at_timepoint_suffix, covariate_suffixes in covariates.items():
        for covariate_suffix in covariate_suffixes:
            cv_dict = {
                'id': [],
                covariate_suffix: [],
                'duration': []
            }
            for row in full_df.to_dict(orient='records'):
                id = row['study_id']
                if id not in base_survival_df['id'].values:
                    continue
                anti_tnf_start_dol_months = base_survival_df[base_survival_df['id'] == id]['start'].iloc[0]
                for timepoint in range(1, 12+1):
                    measurement_dol_months = None
                    try:
                        measurement_dol_months = row[f'timepoint_{timepoint}_{timepoint}_{measured_at_timepoint_suffix}']
                    except KeyError as e:
                        continue
                    if measurement_dol_months is None:
                        continue

                    if measurement_dol_months < anti_tnf_start_dol_months:
                        warn(f'Timepoint {timepoint} covariate {covariate_suffix} measured {anti_tnf_start_dol_months - measurement_dol_months} months before Anti-TNF induction for {id}')

                    measured_value = None
                    try:
                        measured_value = row[f'timepoint_{timepoint}_{covariate_suffix}']
                    except KeyError as e:
                        continue
                        
                    cv_dict['id'].append(id)
                    cv_dict[covariate_suffix].append(measured_value)
                    cv_dict['duration'].append(measurement_dol_months)
                    
            cv = pd.DataFrame(cv_dict)
            df_with_covariates = add_covariate_to_timeline(df_with_covariates, cv, id_col = 'id', duration_col = 'duration', event_col = 'event')

    return df_with_covariates
```

```python
inova_df_with_covariates = add_covariates_to_inova_survival_df(inova_survival_df, inova_metadata_df)
# adjust start and stop times relative to anti-tnf induction
anti_tnf_start_dol_m = inova_survival_df['start']
inova_df_with_covariates['start'] = inova_df_with_covariates['start'] - anti_tnf_start_dol_m
inova_df_with_covariates['stop'] = inova_df_with_covariates['stop'] - anti_tnf_start_dol_m
inova_df_with_covariates.sort_values(by=['id', 'start']).head(1000)
```

```python
struct_schema = pl.Struct([
    pl.Field("id", pl.String),
    pl.Field("event", pl.Boolean),
    pl.Field("duration", pl.Float64),
])

def is_nonresponder(row):
    study_id = row["study_id"]
    starting_age = row["age_at_induction_date"]
    event_occurred, days = None, None
    # switched by 1st follow-up
    if row['change_intherapydrug'] in range(3, 9+1) or row['fu_therapy'] in range(3, 9+1):
        event_occurred = True
        days = (row["age_at_change_intherapy_date"] - starting_age) * 365.25
    # switched between 1st follow-up and 3-5 year follow-up
    elif row['changetherapy_5yr'] in range(3, 9+1):
        event_occurred = row['therapychange_reason'] not in [1, 3]
        days = (row["therapychange_date_one"] - starting_age) * 365.25
    # never switched
    else:
        # patient never switched
        event_occurred = False
        days = (row["age_at_yr_fu_date"] - starting_age) * 365.25
            
    return pl.Struct({'id': study_id, 'event': event_occurred, 'duration':days})

df = df.with_columns(
    pl.struct(pl.all()).map_elements(is_nonresponder, return_dtype=struct_schema).struct.unnest()
)
```
